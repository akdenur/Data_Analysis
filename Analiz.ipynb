```python
import pandas as pd


df = pd.read_csv('kultur_varliklari.csv')

print("Sütun İsimleri:", df.columns.tolist())
print("\nİlk 3 Satır:")
print(df.head(3))
```

```python
import numpy as np


hedef_sutunlar = ['district_name', 'cultural_period', 'lost_artifact']
df_temiz = df[hedef_sutunlar].copy()

df_temiz['district_name'] = df_temiz['district_name'].fillna('BİLİNMİYOR')
df_temiz['cultural_period'] = df_temiz['cultural_period'].fillna('BİLİNMİYOR')
df_temiz['lost_artifact'] = df_temiz['lost_artifact'].fillna('BİLİNMİYOR')

print("--- EN ÇOK KÜLTÜR VARLIĞI OLAN İLK 5 İLÇE ---")
print(df_temiz['district_name'].value_counts().head(5))

print("\n--- KAYIP ESER DURUMU DAĞILIMI ---")
print(df_temiz['lost_artifact'].value_counts())

print("\n--- KÜLTÜREL DÖNEM DAĞILIMI (İLK 5) ---")
print(df_temiz['cultural_period'].value_counts().head(5))
```

```python
# 1. HİPOTEZ: Fatih ilçesindeki kültür varlığı sayısı, diğer tüm ilçelerin toplamından fazladır.
fatih_sayisi = np.sum(df_temiz['district_name'] == 'FATİH')
diger_ilceler_sayisi = np.sum(df_temiz['district_name'] != 'FATİH')

print("--- 1. HİPOTEZ TESTİ: İLÇE DAĞILIMI ---")
print(f"Fatih İlçesi Eser Sayısı: {fatih_sayisi}")
print(f"Diğer İlçelerin Toplam Eser Sayısı: {diger_ilceler_sayisi}")

if fatih_sayisi > diger_ilceler_sayisi:
    print("Sonuç 1: Hipotez DOĞRULANDI. Fatih, diğer tüm ilçelerin toplamından fazla esere sahiptir.\n")
else:
    print("Sonuç 1: Hipotez ÇÜRÜTÜLDÜ. Fatih'in eser sayısı diğerlerinin toplamından azdır.\n")


# 2. HİPOTEZ: Osmanlı dönemi eserlerinin kayıp oranı, Bizans dönemi eserlerinin kayıp oranından daha düşüktür.
osmanli_verisi = df_temiz[df_temiz['cultural_period'] == 'OSMANLI']
bizans_verisi = df_temiz[df_temiz['cultural_period'] == 'BİZANS']

osmanli_kayip = np.sum(osmanli_verisi['lost_artifact'] == 'EVET')
bizans_kayip = np.sum(bizans_verisi['lost_artifact'] == 'EVET')

# Sıfıra bölünme hatasını önlemek için toplamları alıyoruz
osmanli_toplam = len(osmanli_verisi)
bizans_toplam = len(bizans_verisi)

osmanli_kayip_orani = (osmanli_kayip / osmanli_toplam) * 100 if osmanli_toplam > 0 else 0
bizans_kayip_orani = (bizans_kayip / bizans_toplam) * 100 if bizans_toplam > 0 else 0

print("--- 2. HİPOTEZ TESTİ: DÖNEMSEL KAYIP ORANI ---")
print(f"Osmanlı Dönemi Kayıp Oranı: %{osmanli_kayip_orani:.2f} ({osmanli_kayip}/{osmanli_toplam})")
print(f"Bizans Dönemi Kayıp Oranı: %{bizans_kayip_orani:.2f} ({bizans_kayip}/{bizans_toplam})")

if osmanli_kayip_orani < bizans_kayip_orani:
    print("Sonuç 2: Hipotez DOĞRULANDI. Osmanlı eserlerinin kayıp oranı Bizans'a göre düşüktür.")
else:
    print("Sonuç 2: Hipotez ÇÜRÜTÜLDÜ.")
```
