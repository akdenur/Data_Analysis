```python
import pandas as pd

# CSV dosyasını okuyup 'df' adlı değişkene atıyoruz
df = pd.read_csv('bestsellers.csv')
```

```python
# 'Name' sütununu seçiyoruz
kitap_adlari = df['Name']
kitap_adlari.head()
```

```python
# 'User Rating' sütununu seçiyoruz
kullanici_puanlari = df['User Rating']
kullanici_puanlari.head()
```

```python
# 'Author' sütunundaki ilk 8 veriyi getiriyoruz
ilk_sekiz_yazar = df['Author'].head(8)
print(ilk_sekiz_yazar)
```

```python
essiz_turler = df['Genre'].unique()
print(essiz_turler)
```

```python
farkli_yazar_sayisi = df['Author'].nunique()
print(f"Farklı yazar sayısı: {farkli_yazar_sayisi}")
```

```python
ortalama_fiyat = df['Price'].mean()
print(f"Ortalama Fiyat: {ortalama_fiyat}")
```

```python
en_yuksek_10_fiyat = df['Price'].nlargest(10)
print(en_yuksek_10_fiyat)
```

```python
# 'Name' sütunundaki değerlerin frekansını sayıp en yüksek 3 tanesini alıyoruz
en_yaygin_3_kitap = df['Name'].value_counts().head(3)
print(en_yaygin_3_kitap)
```

```python
# Yalnızca 'Author' ve 'User Rating' sütunlarını seçerek yeni bir DataFrame oluşturuyoruz
df_yeni = df[['Author', 'User Rating']]
df_yeni.head()
```

```python
# Yeni DataFrame üzerinde value_counts() kullanarak en çok tekrar eden kombinasyonu buluyoruz
en_yaygin_kombinasyon = df_yeni.value_counts().head(1)
print(en_yaygin_kombinasyon)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Veri setini okuma (eğer önceki hücrelerde okumadıysan)
df = pd.read_csv('bestsellers.csv')

# Türlerin frekansını hesaplayıp pasta grafiği çizdirme
df['Genre'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=['#ff9999','#66b3ff'],
    startangle=90,
    figsize=(8, 8)
)
plt.title('Kitap Türlerinin Dağılımı (Genre)')
plt.ylabel('') # Y ekseni etiketini gizlemek için
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# En çok kitap yazan ilk 10 yazarı bulup çubuk grafik çizdirme
df['Author'].value_counts().head(10).plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)
plt.title('En Üretken 10 Yazar')
plt.xlabel('Yazar')
plt.ylabel('Kitap Sayısı')
plt.xticks(rotation=45, ha='right') # İsimlerin düzgün görünmesi için eğik yazdırıyoruz
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

# Kullanıcı puanlarının dağılımı için histogram çizdirme
df['User Rating'].plot(
    kind='hist',
    bins=15,
    color='lightgreen',
    edgecolor='black'
)
plt.title('Kullanıcı Puanlarının Dağılımı')
plt.xlabel('Kullanıcı Puanı (User Rating)')
plt.ylabel('Frekans (Kitap Sayısı)')
plt.show()
```
