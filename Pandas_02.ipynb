```python
import pandas as pd

df_everest = pd.read_csv('mount_everest_deaths.csv', index_col='No.')

df_everest.head()
```

```python
eksik_degerler = df_everest.isnull().sum()

sifir_eksik = eksik_degerler[eksik_degerler == 0].index.tolist()
print(f"Sıfır eksik değer barındıran sütunlar: {sifir_eksik}")
```

```python
en_cok_eksik_sutun = eksik_degerler.idxmax()
en_cok_eksik_sayisi = eksik_degerler.max()

print(f"En fazla eksik değere sahip sütun: '{en_cok_eksik_sutun}' ({en_cok_eksik_sayisi} eksik değer)")
```
